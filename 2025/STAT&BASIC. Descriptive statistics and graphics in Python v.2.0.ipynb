Descriptive statistics is a formula that uses the entire data set and gives us an answer as a single number.

Measures of central tendency:

- Mean (arithmetic mean of all values).
- Median (middle value).
- Mode (most frequent observation).

Measures of dispersion:
- Variance
- Standard deviation
- Range
- Interquartile range

Random variable - what we work with. It is called random because when we select a number from a data set, we do not know what it will be. 
It is denoted by a capital letter.

Observation - a specific value of the random variable. 
It is denoted by a small letter.

Population - the entire set of objects that we are interested in. 

Sample - since in real life, as a rule, we do not have information about the entire population, we have to work with some part of it. 
This part of the population is called a sample. It is denoted by a small letter.

Important!
We should always remember what we are working with - a sample or a general population. This will affect the calculation method we choose.

```python
import pandas as pd

df = pd.DataFrame({
    'men': [80.2, 80.8, 57.5, 98, 50.5, 73.8, 77.4, 59.7, 77.9, 52.6],
    'women': [57.9, 57.3, 72.8, 48.2, 59.5, 48.3, 61.2, 53.9, 53.9, 70.7]
})

mean_men = df['men'].mean()
mean_women = df['women'].mean()

difference = mean_men - mean_women
rounded_difference = round(difference, 1)
rounded_difference
```

Quartiles divide a dataset into four equal parts (each containing 25% of the data):
- Lower quartile (Q1): Separates the lowest 25% of values from the rest.
- Upper quartile (Q3): Separates the highest 25% of values from the rest.
- Second quartile (Q2): The median of the dataset.

How to find quartiles:
- Sort the dataset in ascending order.
- Find the median (Q2), which divides the data into two halves.
- Find the median of the lower half → this is Q1.
- Find the median of the upper half → this is Q3.

If we calculated quartiles using the inclusive method, we would get different values since each half would have an odd number of elements.

There’s also a third approach, which is a compromise between the two methods:
when the dataset has an odd number of values, you include or exclude the median in the halves in such a way that each remaining half contains an odd number of elements.


```python
import numpy as np

data1 = sorted([2, 7, 4, 3, 6, 4, 1, 2])
data2 = sorted([5, 2, 4, 8, 1, 6, 7])

def exclusive_quartiles(data):
    n = len(data)
    mid = n // 2
    if n % 2 == 0:
        lower_half = data[:mid]
        upper_half = data[mid:]
    else:
        lower_half = data[:mid]
        upper_half = data[mid+1:]
    q1 = np.median(lower_half)
    q3 = np.median(upper_half)
    return q1, q3

q1_data1, q3_data1 = exclusive_quartiles(data1)
q1_data2, q3_data2 = exclusive_quartiles(data2)

q1_data1, q3_data1, q1_data2, q3_data2
```

In pandas, you can calculate quartiles using the quantile() function. The quartile value is specified as a number between 0 and 1 (e.g., Q1 = 0.25, Q3 = 0.75).

The quantile() function has an optional parameter interpolation that defines how the value is calculated when it falls between two data points.

- Default method: interpolation='linear'
- To calculate the quartile as the midpoint between two points, use interpolation='midpoint'.

For this task and the following ones, you should use interpolation='midpoint'.

```python
import pandas as pd

df = pd.DataFrame({
    'men': [80.2, 80.8, 57.5, 98, 50.5, 73.8, 77.4, 59.7, 77.9, 52.6],
    'women': [57.9, 57.3, 72.8, 48.2, 59.5, 48.3, 61.2, 53.9, 53.9, 70.7]
})

q1_men = round(df['men'].quantile(0.25, interpolation='midpoint'), 1)
q3_men = round(df['men'].quantile(0.75, interpolation='midpoint'), 1)

q1_women = round(df['women'].quantile(0.25, interpolation='midpoint'), 1)
q3_women = round(df['women'].quantile(0.75, interpolation='midpoint'), 1)

weight_25_percent_men = round(q3_men, 1)

q1_men, q3_men, q1_women, q3_women, weight_25_percent_men
```

Key points in English:

* **Range**: The difference between the maximum and minimum values in the dataset.
  $\text{Range} = x_{\text{max}} - x_{\text{min}}$

* **Interquartile Range (IQR)**: The difference between the third quartile (Q3) and the first quartile (Q1), representing the spread of the middle 50% of the data.
  $IQR = Q3 - Q1$

* **Standard Deviation (σ)**: A measure of how much the data deviates from the mean.

  * **Population standard deviation**:
    $\sigma = \sqrt{\frac{\sum (x_i - \mu)^2}{n}}$
  * **Sample standard deviation (s)**:
    $s = \sqrt{\frac{\sum (x_i - \bar{x})^2}{n - 1}}$

* **Variance**: The square of the standard deviation, expressed in squared units.

* Dispersion and standard deviation are never negative.
